# 01 — Data Validation & Exploratory Data Analysis
This notebook validates the King-County dataset (duplicates, train/test overlap, coordinates, date parsing) and inspects the target distribution and spatial structure.

```python
%matplotlib inline
# Utilities to add the project root to sys.path so `import src` works from notebooks.
import sys, os
from pathlib import Path
ROOT = Path(os.getcwd()).resolve()
while 'src' not in [p.name for p in ROOT.iterdir()] and ROOT.parent != ROOT:
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
from src.data.load import load_clean_train, load_clean_test

df = load_clean_train()
test = load_clean_test()
print('Clean train:', df.shape, '| test:', test.shape)
print('Duplicate ids in train:', int(df['id'].duplicated().sum()))
print('sale years:', df['sale_year'].min(), '-', df['sale_year'].max())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 4))
sns.histplot(df['price'], bins=60)
plt.title('Price distribution (right-skewed, log used downstream)')
plt.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(7, 6))
sc = plt.scatter(df['long'], df['lat'], c=df['price'], cmap='viridis', s=4)
plt.colorbar(sc, label='price')
plt.title('Geographic price structure (King County, WA)')
plt.tight_layout(); plt.show()
```

### Findings
- 99 duplicate ids are repeat sales of the same home; the most recent is kept.
- 70 ids overlap train/test but describe identical properties; test has no labels so no target leakage.
- Coordinates are all valid; `date` parsed into `sale_year` / `sale_quarter` (2014-2015).
